### Упражнение 2

### Упражнение 3
Напишите программу на Python, которая запрашивает у пользователя два числа и выводит их сумму.

```python
num1 = float(input("Введите первое число: "))
num2 = float(input("Введите второе число: "))
total = num1 + num2
print("Сумма чисел:", total)
```

### Упражнение 4
Напишите программу на Python, которая запрашивает у пользователя три слова и выводит их в обратном порядке, разделяя пробелами.

```python
word1 = input("Введите первое слово: ")
word2 = input("Введите второе слово: ")
word3 = input("Введите третье слово: ")

print(word3, word2, word1)
```

```python
w1 = input("Первое слово: ")
w2 = input("Второе слово: ")
w3 = input("Третье слово: ")

print(w3, w2, w1)
```
